# Object-Oriented Programming (OOP) in Python

## What is OOP?
**Object-Oriented Programming (OOP)** is a programming style where a program is organized around **objects**, which bundle data (what something *has*) and behavior (what something *does*) together, instead of writing everything as a sequence of separate functions.

> **Real-world scenario:** Think of a car factory. Instead of describing every car from scratch, the factory has one **blueprint** and builds many cars from it. Each car has its own color and mileage, but all of them can start, stop, and accelerate.

---

## Class
A **class** is a blueprint or template that defines what data and behavior its objects will have. It does not represent a real thing by itself.

> **Real-world scenario:** The car blueprint. It says every car has a color, brand, and speed, and can start, brake, and accelerate. But the blueprint itself is not a car you can drive.

## Object (Instance)
An **object** is a real, usable thing created from a class. Each object has its own values for the data defined in the class. Creating an object is called **instantiation**.

> **Real-world scenario:** A red Honda parked in your garage, built from the car blueprint. Another blue Toyota built from the same blueprint is a different object with its own values.

## Attribute
An **attribute** is a variable that belongs to an object or class. It stores the object's data or state.

> **Real-world scenario:** A car's color, brand, and current speed.

## Method
A **method** is a function defined inside a class. It describes what an object can do and usually works on the object's own attributes.

> **Real-world scenario:** A car's ability to start, brake, or accelerate.

## Constructor (`__init__`)
The **constructor** is a special method that runs automatically when an object is created. It is used to set the object's initial attribute values.

> **Real-world scenario:** When the factory builds a car, it assigns the color, brand, and engine number at that moment.

## `self`
**`self`** is a reference to the current object. It lets a method access that particular object's own attributes and other methods.

> **Real-world scenario:** When you press "start" in your car, it starts *your* car, not every car in the world.

---

## The 4 Pillars of OOP

| Pillar | Definition | Real-world scenario |
|---|---|---|
| **Encapsulation** | Bundling data and methods together and restricting direct access to the internal data. | You use a car's steering wheel and pedals without touching the engine directly. |
| **Abstraction** | Hiding complex internal details and exposing only what is necessary. | You press the accelerator without knowing how fuel injection works. |
| **Inheritance** | A new class (child) reuses and extends the attributes and methods of an existing class (parent). | An electric car is a kind of car, so it gets all car features plus its own, like a battery. |
| **Polymorphism** | The same method name behaves differently depending on the object. | "Start" works differently for a petrol car, an electric car, and a bike, but you use the same action. |

## Class & Instance Attributes

A class attribute belongs to the class and is shared by all objects. An object attribute belongs to one specific object, so each object has its own copy.

```python
class Studnets:
    collage_name = "IIT Madras" # class attributes
    name = "anonymous"
    def __init__(self, full_name, rollno):
        print(f"hello! {full_name} , your roll no is {rollno}")
        # Insatnce or object attribute
        self.name = full_name
        self.roll_no = rollno


s1 = Studnets("Mukund",1) # Studnets.__init__(s1,"mukund",1)
print(s1.name)
print(s1.roll_no)
print(s1.collage_name)
print(Studnets.collage_name)

# Object attribute  > class attribute
```

```python
class Student:
    school = "IITM"            # class attribute (shared)

    def __init__(self, name):
        self.name = name       # object attribute (separate for each object)

s1 = Student("Dude")
s2 = Student("Rahul")

print(s1.school, s2.school)    # IITM IITM
print(s1.name, s2.name)        # Dude Rahul
```

## Static methods:

- Methods that don't use the self parameter (work at class level)

### Decrorators:
- allow us to wrap another fuction in order to extend the behaviour of the wrapped function, without permanently modifying it.   @staticmethod  --it convert normal method to static method

```python
class Studnets:

    collage = "IIT Madras"
    def __init__(self, full_name, rollno):
        # print(f"hello! {full_name} , your roll no is {rollno}")
        self.name = full_name
        self.roll_no = rollno

    @staticmethod
    def hello(): # idf we don't want to use self then we use static method 
        print("hello new student")
        print(f"welcom to {Studnets.collage}")

s1 = Studnets("Mukund",1) 
s1.hello()
```

```python
# If a function in a class has no self argument and no @staticmethod:
# - it works when called through the class: Student.is_adult(20)
# - it fails when called through an object: s.is_adult(20)
# Reason: Python automatically passes the object as the first argument
# when you call a method through an object.

class Student:
    def is_adult(age):        # no self, no @staticmethod
        return age >= 18

print(Student.is_adult(20))   # True, works through the class

s = Student()
print(s.is_adult(20))         # TypeError: takes 1 positional argument but 2 were given
```

```python
class Student:
    @staticmethod
    def is_adult(age):
        name = "dude"
        return age >= 18

print(Student.is_adult(20))   # True
s = Student()
print(s.is_adult(20))         # True
```

## Important

### Abstarction:
- Hiding the implementation details of a class and only showing the essential features to the user.

### Encapsulation:
- Wrapping data and function into a single unit(object).

```python
class Car:
    def __init__(self):
        self.start = False
        self.brk = False
        self.gear = False

    def car_start(self):
        self.gear= True
        self.start= True
        print(" your care is started let's gooooooooooo")

car1 = Car()
car1.car_start() # so see here we are hiding uncessary details like : steps , start = True .... these stuffs we are only showing print  statemnet so here this is called abstraction
```

### Practicse Question : Create Account class with 2 attributes- balance & account no. Create methods for debit , credit & printing the balance.

```python
class Account:
    def __init__(self, bal , acc):
        self.balance = bal
        self.account_no = acc

    # debit method
    def debit(self, amount):
        self.balance -= amount
        print(f"Rs.{amount} was debited")


    def credit(self, amount):
        self.balance += amount
        print(f"Rs.{amount} was credited")

    def get_balance(self):
        print(f"here is your final balnce in your {self.balance}")


acc1 = Account(20000,1235)
print(acc1.balance)  
print(acc1.account_no)

#debit
acc1.debit(1000)
print(acc1.balance)     
acc1.get_balance() 

print("##############")
#credit
acc1.credit(2000)
print(acc1.balance)
acc1.get_balance()
```
